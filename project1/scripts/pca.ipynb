```python
import numpy as np
import pandas as pd

```

```python
# loading sample metadata
samples = pd.read_csv(
    "../data/raw/original_pf8/Pf8_samples (1).txt",
    sep="\t",
    engine="python"
)
```

```python
samples.shape
```

```python
samples.head()
```

```python
# loading the mean genotypes
dist = np.load(
    "../data/raw/original_pf8/Pf8_mean_genotype_distance_snp_only.npy",
    mmap_mode="r"
)
```

```python
# defining africa countries
african_countries = [
    'Mauritania',
    'Gambia',
    'Guinea',
    'Kenya',
    'Tanzania',
    'Ghana',
    'Burkina Faso',
    'Mali',
    'Malawi',
    'Uganda',
    'Democratic Republic of the Congo',
    'Nigeria',
    'Madagascar',
    'Cameroon',
    "Côte d'Ivoire",
    'Ethiopia',
    'Benin',
    'Senegal',
    'Gabon',
    'Sudan',
    'Mozambique'
]
```

```python
samples.columns
```

```python
# filtering for africa countries that have passed qc 
filtered_samples = samples[
    (samples["QC pass"] == True ) &
    (samples["Country"].isin(african_countries))
]

print(filtered_samples.shape)
filtered_samples.head()
```

```python
# adding sample id to the mean distance matrix
ids = samples["Sample"].tolist()   

dist_df = pd.DataFrame(
    dist,
    index=ids,
    columns=ids
)
```

```python
# filtered african countries id
filtered_ids = filtered_samples["Sample"].tolist()
```

```python
# getting distance mean matrix for African countries
dist_africa = dist_df.loc[filtered_ids, filtered_ids]
```

```python
dist_africa.shape
```

```python
dist_africa.to_csv('dist_africa.csv',index = True)
```

```python
np.save("dist_africa.npy", dist_africa.to_numpy())
```
