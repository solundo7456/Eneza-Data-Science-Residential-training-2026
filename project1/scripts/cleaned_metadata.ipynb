```python
# Filter by country/Region
# First select or the interested countries ie East and West African countries
# Grouped he ountriess as East and West Afrian contries
east_africa = [
    "Kenya",
    "Tanzania",
    "Uganda",
    "Rwanda",
    "Burundi",
    "Ethiopia",
    "South Sudan",
    "Mozambique"
]

west_africa = [
    "Ghana",
    "Nigeria",
    "Mali",
    "Burkina Faso",
    "Senegal",
    "Guinea",
    "Sierra Leone",
    "Liberia",
    "Benin",
    "Togo",
    "Niger",
    "Côte d'Ivoire",
    "The Gambia",
    "Cameroon"   # Include only if your project considers it part of the West/Central comparison
]
```

```python
# list all east African countries 
east_africa = [
    "Kenya",
    "Uganda",
    "Tanzania",
    "Rwanda",
    "Burundi",
    "Ethiopia",
    "South Sudan",
    "Sudan",
    "Somalia",
    "Eritrea",
    "Djibouti"
]

East_Africa_metadata = metadata[metadata["country"].isin(east_africa)]

```

```python
# All the east Afrian countries in out metadata
East_Africa_metadata["country"].value_counts()
```

```python
#list of all thee west african countries
west_africa = [
    "Ghana",
    "Nigeria",
    "Mali",
    "Burkina Faso",
    "Senegal",
    "Guinea",
    "Sierra Leone",
    "Liberia",
    "Benin",
    "Togo",
    "Niger",
    "Côte d'Ivoire",
    "The Gambia",
    "Cameroon"   # Include only if your project considers it part of the West/Central comparison
]

west_Africa_metadata = metadata[metadata["country"].isin(west_africa)]

West_Africa_metadata = metadata[metadata["country"].isin(west_africa)]
```

```python
# List of west african countries in out metadata
west_Africa_metadata["country"].value_counts()
```

```python
# Grouping all the east African countries with their amount of data in each year
east_country_year = pd.crosstab(
    East_Africa_metadata["country"],
    East_Africa_metadata["year"]
)

# Sort years
east_country_year = east_country_year.reindex(
    sorted(east_country_year.columns), axis=1
)

east_country_year
```

```python
# making the data available as csv
east_country_year.to_csv("east-country_year.csv",index= True)
```

```python
# Grouping all the west African countries with their amount of data in each year

west_country_year = pd.crosstab(
    West_Africa_metadata["country"],
    West_Africa_metadata["year"]
)

west_country_year = west_country_year.reindex(
    sorted(west_country_year.columns), axis=1
)

west_country_year
```

```python
# making the data available as csv
west_country_year.to_csv("west-country_year.csv",index= True)
```

```python
# merging east and west Africa that is qrouped yearly
east_year = East_Africa_metadata["year"].value_counts().sort_index()
west_year = West_Africa_metadata["year"].value_counts().sort_index()

regional_year = pd.DataFrame(
    [east_year, west_year],
    index=["East Africa", "West Africa"]
).fillna(0).astype(int)

regional_year
```

```python
# making the data available as csv
regional_year.to_csv("regional_year.csv",index= True)
```

```python
# Filter the years from 2019 to 2019
East_Africa_2010_2019 = East_Africa_metadata[
    East_Africa_metadata["year"].between(2010, 2019)
]
West_Africa_2010_2019 = West_Africa_metadata[
    West_Africa_metadata["year"].between(2010, 2019)
]
```

```python
# Check that every year has at least 71 samples
print(East_Africa_2010_2019["year"].value_counts().sort_index())

print(West_Africa_2010_2019["year"].value_counts().sort_index())
```

```python
# Randomly select 71 samples from each year in East Africa
East_Africa_71 = (
    East_Africa_2010_2019
    .groupby("year", group_keys=False)
    .sample(n=71, random_state=42)
)
```

```python
# # Randomly select 71 samples from each year in west Africa

West_Africa_71 = (
    West_Africa_2010_2019
    .groupby("year", group_keys=False)
    .sample(n=71, random_state=42)
)
```

```python
# Verify
East_Africa_71["year"].value_counts().sort_index()
```

```python
# verify
West_Africa_71["year"].value_counts().sort_index()
```

```python
# East Africa: Total samples per country (across all years)
East_Africa_71["country"].value_counts().sort_index()
```

```python
# West Africa: Total samples per country (across all years)
West_Africa_71["country"].value_counts().sort_index()
```
